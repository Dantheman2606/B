```python
import numpy as np
import matplotlib.pyplot as plt

# Define function
def f(x):
    return (x + 3) ** 2

# Define derivative
def gradient(x):
    return 2 * (x + 3)

# Initial values
x = 2
learning_rate = 0.1
iterations = 50

# Store points for graph
x_values = []
y_values = []

# Gradient Descent
for i in range(iterations):
    x_values.append(x)
    y_values.append(f(x))

    x = x - learning_rate * gradient(x)

# Final minimum
min_x = x
min_y = f(x)

# Print result
print("Minimum Point:")
print("X =", round(min_x, 6))
print("Y =", round(min_y, 6))

# Generate function graph
x_graph = np.linspace(-6, 4, 200)
y_graph = f(x_graph)

# Plot graph
plt.figure(figsize=(9, 5))

plt.plot(x_graph, y_graph, label='y = (x + 3)^2')

# Plot Gradient Descent iterations
plt.scatter(x_values, y_values, color='red',
            label='Gradient Descent Steps')

# Highlight minimum
plt.scatter(min_x, min_y, color='green',
            s=100, label='Minimum Point')



plt.xlabel("X")
plt.ylabel("Y")
plt.title("Gradient Descent")
plt.legend()
plt.grid(True)

plt.show()
```
